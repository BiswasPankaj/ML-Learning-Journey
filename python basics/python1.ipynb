**Python JSON**

JSON is text, written with javascript object notation

```python
#Parse JSON - Convert from JSON to Python -> you can parse json by using json.loads()

import json
from binascii import a2b_hex

#some json
x = '{"name": "Biswas", "age": 30, "city": "New York"}'

#parse x
y = json.loads(x)

#The result is a Python dictionary
print(y["name"])
print(y["age"])
print(y["city"])
```

```python
#convert from Python to JSON -> you can use json.dumps()

import json

x = {
    "name": "Biswas",
    "age": 30,
    "city": "New York"
}

#convert to JSON:
y = json.dumps(x)

#the result is a JSON string:
print(y)
```

```python
#Format the results
# you can use indent to add indents
json.dumps(x, indent=4)

#seperator argument can be used to change default seperator
json.dumps(x, indent = 4, seperator = (". ","= "))

#you can also sort the result
json.dumps(x, indent =4, sort_keys= True)
```

**REGEX**

```python
#it helps use regular expressions and string searches
import re
txt = "The rain in the beautiful month of august makes it india"
x = re.search("^The.*india$", txt)
# the string read -> '^' starts with, '.*' any number of words, '$' ends with
if x:
    print("Match")
else:
    print("Not found")
```

```python
#the findall() function
import re
txt = "The rain in spain"
x = re.findall("ai", txt)
print(x)  #it returns the matches in order, if no match found, empty list returned
```

```python
#The split() function -> returns a list where the string has split at each match

import re
txt = "The rain in spain"
x = re.split(r"\s", txt)
print(x)

```

```python
import re
txt2 = "your lie in april"
y = re.split(r"\s", txt2, maxsplit=2)
print(y)
```

```python
#The sub() function -> replaces thw matches with the text of your choice
import re

txt = "The rain in spain"
x = re.sub(r"\s", "69", txt)
print(x)
```

```python
#you can also control number of replacements

import re
txt = "The rain in spain"
x = re.sub(r"\s", "9", txt, count=2)
print(x)
```

```python
#Match object -> it takes 3 properties, span,string and group
import re

txt = "The rain in Spain"
x = re.search("ai", txt)
print(x) #this will print an object
```

```python
#span() returns a tuple containing the start and end positions of match
import re

txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt)
print(x.span())
print(x)
print(x.string)
print(x.group())
```

```python
#Python self parameter -> reference to current instance of the class, used to access properties and methods that belong to the class

#it doesnt have to be called "self" name it whatever you like, but it has to be the first parameter
class Person:
  def __init__(myobject, name, age):
    myobject.name = name
    myobject.age = age

  def greet(abc):
    print("Hello, my name is " + abc.name)

p1 = Person("Emil", 36)
p1.greet()
```

```python
#calling Methods with self

class Person:
  def __init__(self, name):
    self.name = name

  def greet(self):
    return "Hello, " + self.name

  def welcome(self):
    message = self.greet()
    print(message + "! Welcome to our website.")

p1 = Person("Tobias")
p1.welcome()
```

```python
#the __str__() Method   -> special method that returns what is returned when an object is printed:

#without using __str__()
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)
print(p1)
```

```python
#with __str__()
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} age:({self.age})"

p1 = Person("Tobias", 36)
print(p1)
```

**INHERITANCE**

```python
#Python Inheritance

#allows to define a class that inherits all the methods & properties from another class

class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

x = Person("Biswas", "Pankaj")
x.printname()
```

```python
#Now we create a child class
class Student(Person):
    pass

x = Student("Gauri", "Kaushik")
x.printname()
```

```python
# we can also add the __init__() function to the child class instead of the pass
# when we add __init__() function to the child class, it will not inherit parents __init__() function

class Student(Person):
    def __init__(self, fname, lname):
        #add properties etc.
```

```python
#to keep the inheritance of the parent's __init__() function, add a call to the __init__() function

class Student(Person):
    def __init__(self, fname, lname):
        Person.__init__(self, fname, lname)
```

```python
#use the super() function -> makes the child class inherit all the methods & properties from its parents

class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)    #inherits all methods from parents
        self.graduationyear = year

x = Student("Biswas", "Pankaj", 2023)
print(x.graduationyear)
```

```python
#Python encapsulation -> protecting data inside class

#private properties -> created by adding __
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

p1 = Person("Biswas", 24)
print(p1.name)
print(p1.age)
```

```python
#To access private property value, we can create a getter method

class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    def getage(self):
        return self.__age

p1 = Person("Biswas", 24)
print(p1.getage())
```

```python
#set private property value, for this we create a setter method

class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("age can only be a positive integer")

p1 = Person("Biswas", 18)
print(p1.get_age())

p1.set_age(24)
print(p1.get_age())
```

```python
# Python inner classes
# Accessing inner Class from the outside:

class Outer:
    def __init__(self):
        self.name = "Outer"

    class Inner:
        def __init__(self):
            self.name = "Inner"


        def display(self):
            print("Hello, Inner")

outer = Outer()
inner = outer.Inner()
inner.display()
```

```python
#accessing Outer class from inner class
class Outer:
    def __init__(self):
        self.name = "Biswas"

    class Inner:
        def __init__(self, outer):
            self.outer = outer

        def display(self):
            print(f"Outer class name: {self.outer.name}")

outer = Outer()
inner = outer.Inner(outer)
inner.display()
```

**FILE HANDLING**

python has several func. to create, read, update and delete files

1. "r"- Read -Default value
2. "a"- Append -open a file for appending
3. "w"- write -opens a file for writing
4. "x"- Create -create the specified file

1. "t"- Text -Default value
2. "b"- Binary -Binary mode


```python
#To open a file for reading
f = open("demofile.txt")

f = open("demofile.txt", "rt") #This is the same as above
```

```python
f = open("demofile.txt")
print(f.read())
```

```python
#The with statement -> do not need to worry about closing our files
with open("demofile.txt") as f:
    print(f.read())
```

```python
#although its a good pratice to close files you are done with
f = open("demofile.txt")
print(f.readline())
f.close()
```

```python
#READING ONLY PARTS OF THE FILE
with open("demofile.txt") as f:
    print(f.read(5))  #reads 5 characters
    print(f.readline()) #reads one line
```

```python
#write to an existing File
with open("demofile.txt", "a") as f:
    f.write("I have added this part right now!")

with open("demofile.txt") as t:
    print(t.read())
```

```python
#Overwrite existing content

with open("demofile.txt", "w") as f:
    f.write("I have made a terrible mistake")

with open("demofile.txt") as f:
    print(f.read())
```

```python
#create a new file
f = open("myfile.txt", "x")
```
