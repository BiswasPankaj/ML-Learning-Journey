**DIFFERENCE B/W INTERSECTION and &**

& operator only joins sets with sets, intersection supports other data types

```
#DIFFERENCE
#returns a new set that contains only the items from the first set and not present in other set
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)

print(set3)

#you can also use '-' operator
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 - set2
print(set3)
```

**HOW IS DIFFERENCE AND '-' different?**

'-' operator only works amongst sets, difference() can use different data types

```
#SYMMETRIC DIFFERENCES
#it keeps the only elements that are NOT present in both sets
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2)

print(set3)

#you can also use '^' operator
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 ^ set2
print(set3)
```

similarly, '^' works with only sets, symmetric differences can use diffrent data types

**Python match statement**
instead of writing many if..else statements, you can use match statement.

match expression:\
  case x:\
        code block\
  case y:\
    code block\
  case z:\
    code block

```
#e.g of match
day = 4
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
```

```
#use the underscore character _  as the last case value
day = 4
match day:
    case 6:
        print("saturday")
    case 7:
        print("sunday")
    case _:
        print("i dont know lmao")

#The _ character always match, so make sure to place it as last case to make it acr as a default case
```

```
#Combine values
# use the pipe character | as an operator in the case evaluation to check more than one value
day = 6
match day:
    case 1 | 2 | 3 | 4 | 5:
        print("Great weekday to work!")
    case 6 | 7:
        print("Time to relax!")
```

```
#If statements as guards
#They act as an extra condition check

month = 5
day = 4
match day:
    case 1 | 2 | 3 | 4 | 5 if month == 4:
        print("weekday in april")
    case 1 | 2 | 3 | 4 | 5 if month == 5:
        print("weekday in may")
    case _:
        print("no match found")
```

```
#finding the maximum number
def maxi(*numbers):
    if len(numbers) == 0:
        return None
    max_num = numbers[0]
    for x in numbers:
        if x > max_num:
            max_num = x
    return max_num

print(maxi(23, 45, 12))
```

**Python Decorators**

a decorator lets you add extra behaviour to a function, without changing the functions code\
a decorator is a function that takes another function as input & returns a new function

```
def capital(func):
    def wrapper():
        return func().upper()
    return wrapper

def Lower(func):
    def wrapper():
        return func().lower()
    return wrapper

@Lower
@capital
def myfunc():
    return "HelLo BisWas"

print(myfunc())
```

```
#arguments in the decorator function

def changecase(func):
    def wrapper(x):
        return func(x).upper()
    return wrapper

@changecase
def myfunc(name):
    return "Hello " + name

print(myfunc("Biswas"))
```

```
# *args and *kwargs

#sometimes decorator functions have no control over the arguments passed from decorator functions. to solve this problem, add *args and **kwargs to the wrapper function
#This way the wrapper function can accept any number of arguments

def changecase(func):
    def myinner(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return myinner

@changecase
def myfunc(name):
    return "wassup " + name

print(myfunc("Biswas"))
```

```
#Multiple Decorators

def changecase(func):
    def myinner():
        return func().upper()
    return myinner

def addgreeting(func):
    def myinner():
        return "Hello " + func() + " Have a great day ahead!"
    return myinner

@changecase
@addgreeting
def myfunc():
    return "Biswas"

print(myfunc())
```

```
#Preserving Function Metadata

#functions have metadata you can access using the __name__ and __doc__ attributes

def myfunc():
    return "Have a fantastic day!"

print(myfunc.__name__)
```

```
#remember that when a function is decorated, the metadata of the original function is lost

def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
```

```
#To fix this, python has built-in function called functools.wraps

import functools

def changecase(func):
    @functools.wraps(func)
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunc():
    return "Have a fantastic day!"

print(myfunc.__name__)
```

```
#Lambda with built-in functions --> map(), filter(), sorted()
# map() function applies a function to every item in an iterable
# E.g-> double all numbers in a list

num = [1, 2, 3, 4, 5]
doubled = list(map(lambda X: X * 2, num))
print(doubled)
```

```
#Lambda with filter()
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_num = list(filter(lambda x : x % 2 != 0, numbers))
print(odd_num)
```

```
#Lambda with sorted()
students = [("Emil", 25), ("Biswas", 24), ("Gauri", 21)]
sorted_students = sorted(students, key=lambda x : x[1])
print(sorted_students)
```

```
#sort strings by length
words = ["apple", "pie", "banana", "strawberry"]
sorted_words = sorted(words, key=lambda x : len(x))
print(sorted_words)
```
